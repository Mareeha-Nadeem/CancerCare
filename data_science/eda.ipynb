```python
from sklearn.ensemble import RandomForestClassifier

X = df.drop('Level', axis=1)
y = df['Level']

model = RandomForestClassifier()
model.fit(X, y)
feat_importances = pd.Series(model.feature_importances_, index=X.columns)
feat_importances.sort_values().plot(kind='barh', figsize=(8,6))
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Group by 'Level' and take mean
agg_df = df.groupby('Level').mean()

# Plot heatmap
plt.figure(figsize=(14, 6))  # wider figure for many features
sns.heatmap(agg_df, annot=True, fmt=".1f", cmap='coolwarm', cbar_kws={'label': 'Mean Value'})

# Rotate x-axis labels for readability
plt.xticks(rotation=90)
plt.yticks(rotation=0)  # keep y-axis labels horizontal
plt.title("Mean Values by Level")
plt.ylabel("Level")
plt.xlabel("Features")
plt.tight_layout()  # adjust layout so nothing is cut off
plt.show()

```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Group by 'Level' and take mean
agg_df = df.groupby('Level').mean()

# Transpose for easier plotting: features on x-axis
agg_df_T = agg_df.T

# Plot line chart
plt.figure(figsize=(16, 6))
for level in agg_df_T.columns:
    plt.plot(agg_df_T.index, agg_df_T[level], marker='o', label=level)

plt.xticks(rotation=90)
plt.xlabel("Features")
plt.ylabel("Mean Value")
plt.title("Mean Feature Values by Level")
plt.legend(title="Level")
plt.tight_layout()
plt.show()
```
